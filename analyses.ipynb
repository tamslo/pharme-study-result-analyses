# Comprehension (primary outcome)


## Non-inferiority of comprehension (at result return)


$H_0: \mu_B - \mu_A \leq -\Delta \Leftrightarrow \mu_B \leq \mu_A - \Delta$

$H_1: \mu_B - \mu_A \gt -\Delta \Leftrightarrow \mu_B \gt \mu_A - \Delta$

- $A$: comprehension in pharmacist-led counseling group (control)
- $B$: comprehension in PharMe group (case)
- $\Delta$: 10%, i.e., one point difference between case and control scores
- $\alpha$: 0.025 (one-sided)

:arrow_right: quite conservative setup

:exclamation: please note that we did not control for multiple testing for
secondary outcomes, interpret results as tendencies


```python
%reload_ext autoreload
%autoreload 2

from modules.analyses.comprehension import analyze_comprehension
from modules.definitions.constants import COMPREHENSION_REMOVED_QUESTIONS
from modules.definitions.types import TimePoint
from modules.utils.comparisons import analyze_time_point_differences

wrong_answers = {}
plot_data = {}
for time_point in TimePoint:
    current_wrong_answers, current_plot_data = analyze_comprehension(
        time_point,
        drop_questions=COMPREHENSION_REMOVED_QUESTIONS,
    )
    wrong_answers[time_point.name] = current_wrong_answers
    plot_data[time_point] = [current_plot_data]
analyze_time_point_differences(Survey.COMPREHENSION, plot_data, dry_run=False)
```

# Cost Estimation

We only consider running costs, neither the initial app development costs, nor
the initial costs to create the counseling program.

For PharMe, running costs include a server running the backend, generously
estimated with $ 300 per month, and the monthly salary of one half-time
software developer.

The counseling cost is mostly the counselors' salary.

Therefore, we assume that PharMe costs less as soon as more than one full-time
counselor is needed.

_Of course this assumes that PharMe replaces a counselor which will likely_
_not be the case; but as soon as PharMe can return as many results without_
_a counseling follow-up as one counselor would have sessions with, PharMe_
_can be assumed as cost-effective._
_This still needs to be shown in further studies, but we are positive that such_
_a scaling can be met._
